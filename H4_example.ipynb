```python
n=8 #number of qubits
```

```python
#This is just used to load the state and Hamiltonian, assuming the script, hamiltonian, and ansatz files are all in the same directory. Modify cell accordingly.
filename='h4'
from pathlib import Path
script_dir = Path().resolve()
H=openfermion.utils.load_operator(file_name='H_'+filename+'2a.data',
    data_directory=str(script_dir), plain_text=True)

spstate=scipy.sparse.load_npz(filename+'2astate.npz')
state=qiskit.quantum_info.Statevector(spstate.toarray())


circuit = qiskit.QuantumCircuit(n)
circuit.initialize(state)
```

```python
g=poolgradients(H, poolgroup('Y', 0, n), circuit, spstate, n, [0.4, 1])
g.measure_gradients_sorted(0.001)
g.gradients_first_eval()
```

```python
g=poolgradients(H, poolgroup('Y', 0, n), circuit, spstate, n, [0.4, 1])
g.measure_hterms_first(poolgroup('Y', 0, n), 0.001)
g.gradients_first_eval()
```

```python
g=poolgradients(H, poolgroup('Y', 0, n), circuit, spstate, n, [0.4, 1])
g.measure_gradients_first_independent(0.001)
g.gradients_first_eval()
```

```python
g=poolgradients(H, poolgroup('Y', 0, n), circuit, spstate, n, [0.4, 1])
g.measure_energy_independent(0.001)
g.energy_eval()
```
